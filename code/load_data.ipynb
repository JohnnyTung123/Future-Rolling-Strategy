```python
import numpy as np
from datetime import datetime
import pandas as pd
import yfinance as yf
```

```python
# ticker = "^GSPC"
tickers_members = pd.read_excel('../data/raw/spx_members.xlsx')
tickers_members
```

```python
tickers = tickers_members[['tickers']].values
tickers = tickers.flatten().tolist()
print(tickers)
```

```python
end_date = datetime.today()
start_date = end_date - pd.DateOffset(years=10)

data = yf.download(
    '^SPX',
    start=start_date,
    end=end_date,
    auto_adjust=True
)
```

```python
data.tail()
```

```python
data['Open'].to_excel("open.xlsx")
data['High'].to_excel("high.xlsx")
data['Low'].to_excel("low.xlsx")
data['Close'].to_excel("close.xlsx")
data['Volume'].to_excel("volume.xlsx")
```
