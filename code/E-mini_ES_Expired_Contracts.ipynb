```python
import numpy as np
from datetime import datetime
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
file_path = '../data/raw/E-mini_ES_hist.xlsx'
years = range(1, 7)
months = ['H', 'M', 'U', 'Z']

contracts = [f'ES{m}{y}' for y in years for m in months]

end_contracts = 'ESH6'

contracts = contracts[:contracts.index(end_contracts)+1]
print(contracts)
```

```python
all_contracts = {}
for contract in contracts:
    df_raw = pd.read_excel(file_path, sheet_name=contract)
    price_df = df_raw[['Date', 'Last Price']].copy()
    volume_df = df_raw[['Date.1', 'Volume']].copy().rename(columns={'Date.1' : 'Date'})

    price_df['Date'] = pd.to_datetime(price_df['Date'])
    volume_df['Date'] = pd.to_datetime(volume_df['Date'])
    price_df = price_df.dropna(subset=['Date'])
    volume_df = volume_df.dropna(subset=['Date'])
    price_df = price_df.set_index('Date')
    volume_df = volume_df.set_index('Date')
    price_df = price_df.sort_index(ascending=True)
    volume_df = volume_df.sort_index(ascending=True)

    df_clean = pd.concat([price_df, volume_df], axis=1)
    all_contracts[contract] = df_clean
    print(f'{contract} cleaned')
    print(df_clean)
```

```python
E_mini_hist = pd.concat(all_contracts, axis=1, sort=True)
E_mini_hist = E_mini_hist.swaplevel(axis=1)
E_mini_hist
```

```python
fut_price = E_mini_hist['Last Price']
fut_price
```
